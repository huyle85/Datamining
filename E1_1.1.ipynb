```python
#Cau 3  Các kiểu dữ liệu nâng cao: List, Tupble, Set, Dictionary(Mỗi kiểu dữ liệu cho mô tả ngắn, ngữ cảnh áp dụng, và ví dụ minh hoạ)

#List
#Mô tả: danh sách các phần tử có thứ tự, có thể trùng lặp giá trị của phần tử
#Ngữ cảnh áp dụng: cần lưu 1 danh sách như là danh sách sinh viên, mảng dữ liệu có thể thay đổi...
#Ví dụ:
ds_sinh_vien = ["Huy", "Dung", "Hùng"]
print("List ban đầu: ", ds_sinh_vien)
ds_sinh_vien.append("Dũng")        
ds_sinh_vien[1] = "Đạt"            
print("List sau khi sửa:", ds_sinh_vien)
print("________________________________________________")


#Tuple
#Mô tả: nó là 1 danh sách cố định, cho phép trùng lặp giá trị của phần tử nhưng không thể thay đổi, cú pháp là để phần tử trong (), bên trong có dấu ,
#Ngữ cảnh áp dụng: luu tọa độ gps, ngày tháng hoặc là những dữ liệu cố định
#Ví dụ:
toa_do = (11.5333, 106.9000)
print("Tọa độ Binh Phuoc:", toa_do)
print("Vĩ độ:", toa_do[0])
try:
    toa_do[1] = 19.000 #thử thay đổi giá trị trong Tuple
except TypeError as errol:
    print(errol)
print("________________________________________________")


#Set
#Mô tả: tập hợp các phần tử không trùng lặp, không có thứ tự và không thể truy cập qua chỉ số
#Ngữ cảnh áp dụng: lọc phần tử trung, tập hợp...
#Ví dụ:
id = {101, 102, 103, 101, 102} 
print("Set sau khi lọc trùng:", id)
id.add(109)
print("Set sau khi thêm 109:", id)
print("________________________________________________")


#Dictionary
#Mô tả: tập hợp lưu theo kiểu key value, key là duy nhất, ứng với key là value, có thể truy cập thông tin qua key
#Ngữ cảnh áp dụng: lưu hồ sơ người dùng, cấu hình Json...
#Ví dụ:
sinh_vien = [
{
    "mssv": "702",
    "ho_ten": "Le Van Huy",
    "diem_tb": 9.5
},
{
    "mssv": "703",
    "ho_ten": "Le Van Heo",
    "diem_tb": 5
}
]
sinh_vien[0]["email"] = "vanwhuy60@gmail.com" 
sinh_vien[0]["diem_tb"] = 9.0          
print("Thông tin sinh viên:", sinh_vien[1])
mssv_find = "702"
for sv in sinh_vien:
    if sv["mssv"] == mssv_find:
        data = sv
        print(data)
```

```python
#Cau 4 Cấu trúc lệnh trong python
#4a Lệnh điều kiện

diem = 8

if diem >= 8.5:
    hocluc = "Xuất sắc"
elif diem >= 7.0:
    hocluc = "Khá"
elif diem >= 5.0:
    hocluc = "Trung bình"
else:
    hocluc = "Yếu"

print(f"Điểm: {diem} | Học lực: {hocluc}")
print("________________________________________________")
#4b vong lap
animals = ["Cho", "Meo", "Chuot"]
for i in animals:
    print("Động vật:", i)

dem = 0
while dem < 5:
    dem += 1
    if dem == 1:
        continue 
    if dem == 5:
        break    
    print("Số đếm:", dem)
```

```python
#Cau 5 Khai báo hàm
# Khai báo hàm tính điểm trung bình và xếp loại
def avg(math, biology, english): 
    dtb = (math + biology + english) / 3
    return dtb
dtb_sv = avg(8.0, 9.0, 8.5)
print("Điểm TB SV:", dtb_sv)
```
