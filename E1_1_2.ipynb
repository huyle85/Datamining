```python
from collections import Counter
file_name = "SMSSpamCollection"
```

```python
#Cau 1 Đếm xem có bao nhiêu dòng và bao nhiêu dòng xuất hiện spam và bao nhiêu dòng xuất hiện ham trong file trên
total_line = 0
spam_count = 0
ham_count = 0
with open(file_name, "r", encoding="utf-8") as f:
    for line in f:
        total_line += 1
        if line.startswith("spam"):
            spam_count += 1
        elif line.startswith("ham"):
            ham_count += 1
print(f"Số dòng: {total_line}")
print(f"Số dòng xuất hiện spam: {spam_count}")
print(f"Số dòng xuất hiện ham: {ham_count}")
```

```python
#Cau 2 Đếm xem có bao nhiêu từ (các từ ngăn cách bởi khoảng trắng, và không phải chữ số)
array_words = []
with open(file_name, "r", encoding="utf-8") as f:
    data = f.read().lower()
words = data.split()
for word in words:
    clean_words = word.strip("*&#$!?:;\"'()[]%@-_/\\=+<>.,{}")
    if clean_words and not clean_words.isdigit():
        array_words.append(clean_words)
print(f"Số từ được ngăn cách bởi khoảng trắng và không phải chữ số: {len(array_words)}")
```

```python
#Cau 3 Đếm xem có bao nhiêu từ 'you' (không phân biệt hoa thường)

count_you = array_words.count("you")
print(f"Số từ 'You': {count_you}")
```

```python
#Cau 4 Đếm xem có bao nhiêu từ khác nhau (không phân biết hoa thường)
count_word = set(array_words)
print(f"Số từ khác nhau là: {len(count_word)}")
```

```python
#Cau 5 Liệt kê 10 từ khóa xuất hiện nhiều nhất trong file dữ liệu (không phân biết hoa thường)
word_count = Counter(array_words)
data = word_count.most_common(10)
for word, count in data:
    print(f"Word: {word} | Count: {count}")
```
