```python
#Cau 1  Đọc file csv và hiển thị dữ liệu
import pandas as pd

url_Iris = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
name_columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'name_flower']
data = pd.read_csv(url_Iris, header=None, names=name_columns)
data.head() 
```

```python
#Cau 2 Cho biết kiểu dữ liệu các cột
data.info()
```

```python
#Cau 3 Hiển thị các giá trị thống kê (count, mean, std,...)
data.describe()
```

```python
#Cau 4 Vẽ biểu đồ histogram cho các cột dữ liệu số
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
data['sepal_length'].plot(kind='hist', bins=15, edgecolor='black', color='skyblue')
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
data['sepal_width'].plot(kind='hist', bins=15, edgecolor='black', color='skyblue')
plt.title('Distribution of Sepal Width')
plt.xlabel('Sepal Width (cm)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
data['petal_length'].plot(kind='hist', bins=15, edgecolor='black', color='skyblue')
plt.title('Distribution of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')
plt.show()  
```

```python
plt.figure(figsize=(8, 5))
data['petal_width'].plot(kind='hist', bins=15, edgecolor='black', color='skyblue')
plt.title('Distribution of Petal Width')
plt.xlabel('Petal Width (cm)')  
plt.ylabel('Frequency')
plt.show()
```
